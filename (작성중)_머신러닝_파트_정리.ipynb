# 용어
### 머신러닝
- 규칙을 프로그래밍하지 않아도 데이터에서 규칙을 자동으로 학습

### 모델
- 머신러닝 알고리즘을 구현한 프로그램 또는 알고리즘을 구체화하여 표현한 것

### 샘플링 편향(sampling bias)
- 훈련 데이터와 테스트 데이터에는 클래스가 한쪽으로 치우치지 않고 골고루 섞여있어야 함

### 데이터 전처리(data preprocessing)
- 특성의 값이 놓인 범위가 크게 다르면 알고리즘이 예측하기 어려우므로 특성값을 일정한 기준으로 맞춰야 함

### 표준 점수, z 점수(standard score)
- 각 특성이 평균에서 표준편차의 몇 배만큼 떨어져 있는지 의미
- 데이터를 표준점수로 변환하는 과정 = 표준화(standardization)
$$
\frac{데이터 - 평균}{표준편차}
$$

### 모델 파라미터
- 머신러닝 알고리즘이 찾은 값, 선형 회귀에서 계수와 절편이 여기에 해당

### 하이퍼파라미터
- 규제 강도와 같이 머신러닝 모델이 데이터를 통해 학습할 수 없고 사람이 설정해야 하는 값

### 특성 공학
- 기존의 특성을 사용해 새로운 특성을 뽑아내는 작업
- (예시) 길이와 높이를 곱해서 새로운 특성을 만듦

### 점진적/온라인 학습
- 이전에 훈련한 모델을 버리지 않고 새로운 데이터에 대해서만 조금씩 더 훈련

# 개념
### k-최근접 이웃 분류
- 분류: 샘플을 몇 개의 클래스 중 하나로 분류 `sklearn.neighbors.KNeighborsClassifier()`
- 가장 가까운 직선 거리에 어떤 데이터가 있는지 살피고 다수를 차지하는 것을 정답으로 판단하는 알고리즘
- 규칙을 찾기보다는 메모리에 데이터를 갖고 있는 것이 특징

### k-최근접 이웃 회귀
- 회귀: 임의의 어떤 숫자를 예측 `sklearn.neighbors.KNeighborsClassifier()`
- 최근접 이웃 샘플 k개를 선택하고 각 샘플의 target 값을 평균하여 예측
- score(input, target) -> 결정계수 반환

### 결정계수(R^2) (coefficient of determination)
$$
R^{2} = 1 - \frac{(target - prediction)^{2} 합}{(target - mean)^{2} 합}
$$
- 분모와 분자가 비슷하다 = 평균 정도 예측했다 = 1 - (1에 가까운 수) = 결정계수는 0에 가까워진다. => 예측을 못 했다.  
- 분자가 0에 가깝다 = 정답에 가깝게 예측했다 = 1 - (0에 가까운 수) = 결정계수는 1에 가까워진다. => 예측을 잘 했다.

### 선형 회귀 (linear regression)
- 특성을 기준으로 예측값에 대한 직선을 학습하는 알고리즘 `sklearn.linear_model.LinearRegression()`
- `lr.coef_`: 직선을 그리기 위한 기울기(계수(coefficient), 가중치(weight))
- `lr.intercept_`: 직선을 그리기 위한 절편
- (예시) 무게 = a * 길이 + b

### 다항 선형 회귀
- 직선대로 예측하면 무게가 0이 되는 것과 같이 틀린 예측을 할 수 있음
- 최적의 곡선을 찾는 경우 보다 정확하게 예측 가능
- 2차 방정식(비선형)이 되지만 이는 입력과 타겟의 관계이고, 가중치와 타겟의 관계가 선형이므로 선형 회귀로 봄
- (예시) 무게 = a * 길이^2 + b * 길이 + c

### 다중 회귀 (multiple regression)
- 특성이 1개면 직선/곡선을 학습, 특성이 2개면 평면을 학습 `sklearn.linear_model.LinearRegression() # 여러 개의 특성을 사용해 선형 회귀`

### 규제 (regularization)
- 머신러닝 모델이 훈련 세트를 과도하게 학습하지 못하도록 규제
- 선형 회귀의 경우, 특성에 곱해지는 계수의 크기를 작게 만들어서 보편적인 패턴 학습
- ridge: 계수(가중치)를 제곱한 값을 기준으로 규제
- lasso: 계수의 절댓값을 기준으로 규제, 아예 0으로 만들 수 있으므로 주요 특성만 추출할 때 사용

### 로지스틱 회귀 (logistic regression)
- 이름은 회귀이지만 분류 모델로 선형 회귀와 동일하게 선형 방정식 학습 `sklearn.linear_model.LogisticRegression()`
- 이진/다중분류 모두 가능
- 확률로 변환하기 위해 시그모이드(로지스틱) 함수를 사용

### 시그모이드 함수
$$\sigma(x) = \frac{1}{1 + e^{-x}}$$
- z가 무한하게 큰 음수이면 0에 가까워지고, 무한하게 큰 양수이면 1에 가까워짐, 0이면 0.5
- 0~1 사이의 값이므로 확률로 해석이 가능해짐

### 다중 분류 (multiclass classification)
- 타깃 데이터에 2개 이상의 클래스가 포함된 문제
- (예시) 데이터가 주어지고 7개 생선 종류 중 어떤 생선에 해당하는지를 맞추기

### 소프트맥스 함수
$$\sigma(z_i) = \frac{e^{z_i}}{\sum_{j=1}^{K} e^{z_j}}$$
- 시그모이드 함수는 선형 방정식 출력값을 0~1 사이로 압축
- 여러 개의 선형 방정식의 출력값을 0~1 사이의 확률로 압축하고 전체 합이 1이 되도록 만듦
- 최댓값에 높은 확률을 할당하면서도 다른 값에도 작은 확률을 부여

### 경사 하강법 (gradient Descent)
- 에포크(epoch): 훈련 세트를 한 번 모두 사용하는 과정
- 확률적(Stochastic) 경사 하강법: 샘플 1개씩 경사 이동 `sklearn.linear_model.SGDClassifier(`
- 미니배치(minibatch) 경사 하강법: 여러 개의 샘플을 사용
- 배치(batch) 경사 하강법: 전체 샘플 이용 -> 컴퓨터 자원을 많이 소모

### 손실 함수 (loss function)
- 머신러닝 알고리즘이 잘못 예측하는지를 측정하는 기준, 작을수록 좋음
- 샘플 하나에 대한 손실
- 로지스틱(이진 크로스엔트로피) 손실 함수
  - $$ L(y, \hat{y}) = -\left[ y \log(\hat{y}) + (1-y) \log(1-\hat{y}) \right] $$
  - -를 앞에 붙이는 이유는 뒤의 계산결과가 클수록 값을 작게 해서 손실이 작도록 하기 위함

### 결정 트리 (decision Tree)
- 이유를 설명하기 쉽다. 테스트 조건, 불순도, 총 샘플 수, 클래스별 샘플 수를 볼 수 있어서 모델 설명 가능

### 지니 불순도 (gini impurity)
- 동일한 분포에서 가져온 새 데이터를 잘못 분류할 확률
- $$I = 1 - (p^{2} + q^{2}) = 1 - (p^{2} + (1-p)^{2})$$
- (예시) $$I = 1 - (0.25^{2} + 0.75^{2}) = 1 - (0.25^{2} + (1-0.75)^{2})$$
  - 동일한 데이터 세트의 무작위 라벨이 잘못 분류될 확률은 37.5%, 올바르게 분류될 확률은 62.5%

### 정보 이득(획득) (information gain)
- 정보 이득이 최대가 되도록 나누는 것이 결정 트리의 알고리즘
- 엔트로피: 확률 분포가 얼마나 예측 불가능한지
- 샘플의 40% 가 있는 하위노드의 엔트로피 0.2, 샘플의 60% 가 있는 하위노드의 엔트로피 0.1
- 상위 노드 엔트로피 - (0.4*0.2 + 0.6*0.1) = 정보 획득

### 비용 함수 (cost function)
- 훈련 세트에 있는 모든 샘플에 대한 손실 함수의 합
