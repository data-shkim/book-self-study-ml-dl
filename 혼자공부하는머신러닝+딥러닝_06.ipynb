```python
import numpy as np
import matplotlib.pyplot as plt

fruits = np.load('fruits_300.npy')
```

```python
print(fruits.shape) # 300개의 샘플, 100*100 크기의 이미지
print(fruits[0,0,:])
```

```python
plt.imshow(fruits[0], cmap='gray') # 흰 바탕 -> 검은 바탕, 검은 사과 -> 흰 사과 => 검은색은 0에 가깝고, 흰색은 255에 가까우므로 반전
plt.show()
```

```python
plt.imshow(fruits[0], cmap='gray_r')
plt.show()
```

```python
fig, axs = plt.subplots(1,2)
axs[0].imshow(fruits[100], cmap='gray_r')
axs[1].imshow(fruits[200], cmap='gray_r')
plt.show()
```

```python
apple = fruits[0:100].reshape(-1, 100*100)
pineapple = fruits[100:200].reshape(-1, 100*100)
banana = fruits[200:300].reshape(-1, 100*100)

print(apple.shape)  # apple 샘플의 개수는 100개, 100*100
```

```python
print(apple.mean(axis=1)) # axis=0 행방향, axis=1 열방향
```

```python
plt.hist(apple.mean(axis=1), alpha=0.8, label='apple')
plt.hist(pineapple.mean(axis=1), alpha=0.8, label='pineapple')
plt.hist(banana.mean(axis=1), alpha=0.8, label='banana')

plt.legend()
plt.show()
```

```python
# 픽셀 평균 구하기 (100개의 샘플에서 0~9999번째의 픽셀 각 평균)
fig, axs = plt.subplots(1, 3, figsize=(20, 5))
axs[0].bar(range(10000), apple.mean(axis=0))
axs[1].bar(range(10000), pineapple.mean(axis=0))
axs[2].bar(range(10000), banana.mean(axis=0))
plt.show()
```

```python
apple_mean = apple.mean(axis=0).reshape(100, 100) # 10000개의 각 픽셀을 -> 100*100 이미지로 변환
pineapple_mean = pineapple.mean(axis=0).reshape(100,100)
banana_mean = banana.mean(axis=0).reshape(100, 100)
fig, axs = plt.subplots(1, 3, figsize=(20, 5))
axs[0].imshow(apple_mean, cmap='gray_r')
axs[1].imshow(pineapple_mean, cmap='gray_r')
axs[2].imshow(banana_mean, cmap='gray_r')
plt.show()
```

```python
abs_diff = np.abs(fruits - banana_mean) # 300 * 100 * 100
abs_mean = np.mean(abs_diff, axis=(1,2))
print(abs_mean.shape)
```

```python
banana_index = np.argsort(abs_mean)[:100]
banana_index = banana_index.reshape(10, 10)

fig, axs = plt.subplots(10, 10, figsize=(10, 10))
for i in range(10):
  for j in range(10):
    axs[i, j].imshow(fruits[banana_index[i, j]], cmap='gray_r')
    axs[i, j].axis('off')
plt.show()
```

```python
import numpy as np

fruits = np.load('fruits_300.npy')
fruits_2d = fruits.reshape(-1, 100*100)
```

```python
from sklearn.cluster import KMeans
km = KMeans(n_clusters=3, random_state=42)
km.fit(fruits_2d)
```

```python
print(km.labels_)
print(np.unique(km.labels_, return_counts=True))
```

```python
import matplotlib.pyplot as plt

def draw_fruits(arr, ratio=1):
  n = len(arr)
  rows = int(np.ceil(n/10))
  cols = n if rows < 2 else 10
  fig, axs = plt.subplots(rows, cols, figsize=(cols*ratio, rows*ratio), squeeze=False)

  for i in range(rows):
    for j in range(cols):
      if i*10 + j < n:
        axs[i, j].imshow(arr[i*10 + j], cmap='gray_r')
      axs[i, j].axis('off')
  plt.show()
```

```python
draw_fruits(fruits[km.labels_ == 0])
```

```python
draw_fruits(fruits[km.labels_ ==1])
```

```python
draw_fruits(fruits[km.labels_==2])
```

```python
draw_fruits(km.cluster_centers_.reshape(-1, 100, 100), ratio=3)
```

```python
print(km.transform(fruits_2d[100:101]))
```

```python
print(km.predict(fruits_2d[100:101]))
```

```python
draw_fruits(fruits[100:101])
print(km.n_iter_)
```

```python
inertia = []
for k in range(2, 7):
  km = KMeans(n_clusters=k, n_init='auto', random_state=42)
  km.fit(fruits_2d)
  inertia.append(km.inertia_)

plt.plot(range(2, 7), inertia)
plt.xlabel('k')
plt.ylabel('inertia')
plt.show()
```

```python
!wget https://bit.ly/fruits_300_data -O fruits_200.npy

import numpy as np
fruits = np.load('fruits_300.npy')
fruits_2d = fruits.reshape(-1, 100*100)
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=50)
pca.fit(fruits_2d)
```

```python
print(pca.components_.shape)
```

```python
draw_fruits(pca.components_.reshape(-1, 100, 100))
```

```python
print(fruits_2d.shape)
```

```python
fruits_pca = pca.transform(fruits_2d)
print(fruits_pca.shape)
```

```python
fruits_inverse = pca.inverse_transform(fruits_pca)
print(fruits_inverse.shape)
```

```python
fruits_reconstruct = fruits_inverse.reshape(-1, 100, 100)
for start in [0, 100, 200]:
  draw_fruits(fruits_reconstruct[start:start+100])
  print("\n")
```

```python
print(np.sum(pca.explained_variance_ratio_))
```

```python
plt.plot(pca.explained_variance_ratio_)
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()

target = np.array([0]*100 + [1]*100 + [2]*100)
```

```python
from sklearn.model_selection import cross_validate
scores = cross_validate(lr, fruits_2d, target)
print(np.mean(scores['test_score']))
print(np.mean(scores['fit_time']))
```

```python
scores = cross_validate(lr, fruits_pca, target)
print(np.mean(scores['test_score']))
print(np.mean(scores['fit_time']))
```

```python
pca = PCA(n_components=0.5)
pca.fit(fruits_2d)
```

```python
print(pca.n_components_)
```

```python
fruits_pca = pca.transform(fruits_2d)
print(fruits_pca.shape)

scores = cross_validate(lr, fruits_pca, target)
print(np.mean(scores['test_score']))
print(np.mean(scores['fit_time']))
```

```python
from sklearn.cluster import KMeans
km = KMeans(n_clusters=3, random_state=42)
km.fit(fruits_pca)
print(np.unique(km.labels_, return_counts=True))
```

```python
for label in range(0, 3):
  draw_fruits(fruits[km.labels_ == label])
  print("\n")
```

```python
for label in range(0, 3):
  data = fruits_pca[km.labels_ == label]
  plt.scatter(data[:,0], data[:,1])
plt.legend(['apple', 'banana', 'pineapple'])
plt.show()
```
